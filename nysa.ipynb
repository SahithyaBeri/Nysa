```python
import speech_recognition as sr
import pyttsx3
import pywhatkit
import datetime
import wikipedia
import pyjokes
import os
import sys
import pyautogui
import webbrowser
import time
```

```python
engine = pyttsx3.init()
engine.setProperty('rate', 170)
voices = engine.getProperty('voices')
engine.setProperty('voice', voices[1].id)
engine.setProperty('voice', voices[0].id)
```

```python
def talk(text):
    print("Nysa:", text)
    engine.say(text)
    engine.runAndWait()
```

```python
def take_command():
    listener = sr.Recognizer()
    with sr.Microphone()as source:
        print("Listening")
        listener.adjust_for_ambient_noise(source)
        voice = listener.listen(source)
    try:
        command = listener.recoginize_google(voice)
        command = command.lower()
        print("You Said:", command)
    except sr.UnknownValueError:
        talk("Sorry Bby, I didn't catch that")
        return""
    except sr.RequestError:
        talk("Network issue with Google service.")
        return ""
    return command  
    
```

```python
def send_whatsapp():
    talk("Who do you want to message?")
    number = input("Enter phone number with country code (e.g. +919876543210): ")
    talk("What should I say?")
    message = take_command()
    talk("Sending your message...")
    pywhatkit.sendwhatmsg_instantly(number, message, 15, True)

def take_screenshot():
    talk("Taking screenshot ")
    screenshot = pyautogui.screenshot()
    screenshot.save("screenshot.png")
    talk("Screenshot saved as screenshot.png")

def open_website():
    talk("Which website should I open?")
    site = take_command().replace("open", "").strip().replace(" ", "")
    if not site.startswith("http"):
        site = "https://" + site
    talk(f"Opening {site}")
    webbrowser.open(site)

def read_calendar():
    talk("Sorry, calendar reading needs Google API setup. Would you like me to help you configure that later?")
```

```python
def take_command():
    listener = sr.Recognizer()
    with sr.Microphone() as source:
        print(" Listening...")
        listener.adjust_for_ambient_noise(source, duration=1)
        voice = listener.listen(source)

    try:
        command = listener.recognize_google(voice)
        command = command.lower()
        print(f" You said: {command}")
    except sr.UnknownValueError:
        print(" I couldn’t understand what you said.")
        command = ""
    except sr.RequestError:
        print(" Could not reach Google service.")
        command = ""
    
    return command
```

```python
talk("Yo! I'm Nysa, your personal voice assistant")

try:
    while True:
        run_nysa()
except KeyboardInterrupt:
    talk("Goodbye! Stopping the assistant.")

```
